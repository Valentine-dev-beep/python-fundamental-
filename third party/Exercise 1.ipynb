```python
import pandas as pd
import matplotlib.pyplot as plt

#loading excel file using read_excel functions
dataframe1 = pd.read_excel('Data.xlsx',sheet_name='Sheet1')
#print(dataframe1)

#get the balance
dataframe1['Balance'] = dataframe1['Car price']- dataframe1['Amount paid'] 
print(dataframe1)


#create a bar chart
plt.figure(figsize=(8, 6))
plt.bar(dataframe1['Vehicle'], dataframe1['Balance'], color='green')
plt.xlabel('Vehicle')
plt.ylabel('Balance')
plt.title('Vehicle Balance')
plt.show()



```
