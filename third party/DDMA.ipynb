```python
"""
December 2024 
question 22
"""
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

# (a)(i) Compute mean and standard deviation
A = np.array([5,6,9,8,7,3,18,3,7,5,14,13,9,7])
mean_A = np.mean(A)
std_A = np.std(A)
print(f"Mean: {mean_A}")
print(f"Standard Deviation: {std_A}")

# (a)(ii) Compute t-test and p-values
t_stat, p_value = stats.ttest_1samp(A, 0)
print(f"T-test statistic: {t_stat}")
print(f"P-value: {p_value}")

# (b) Interpolation and plotting
U = np.array([1,4,6,9])
V = np.array([3,1,2,5])

# Create interpolation function
f = interp1d(U, V, kind='cubic')

# Define interpolation values
U_new = np.linspace(1, 9, 100)
V_new = f(U_new)

# Plot results
plt.figure(figsize=(10, 6))
plt.plot(U, V, 'o', label='Original data')
plt.plot(U_new, V_new, '-', label='Interpolated data')
plt.xlabel('U')
plt.ylabel('V')
plt.title('Interpolation of Data Points')
plt.legend()
plt.grid(True)
plt.show()
```

```python
"""
Question 22

"""
import pandas as pd
import matplotlib.pyplot as plt

# (a)(i) Create dictionary and DataFrame
staff = {
    'Name': ['Dan', 'Anita', 'Susan', 'Caleb'],
    'Age': [25, 30, 35, 40],
    'Salary': [50000, 60000, 55000, 80000]
}
emp = pd.DataFrame(staff)
print(emp)

# (a)(ii) Display specific rows
print("\nSecond row:")
print(emp.iloc[1])
print("\nRow where name is Susan:")
print(emp[emp['Name'] == 'Susan'])

# (a)(iii) Add gender column
emp['Gender'] = ['M', 'F', 'F', 'M']
print("\nDataFrame with Gender:")
print(emp)

# (a)(iv) Rename column
emp = emp.rename(columns={'Age': 'Years'})
print("\nDataFrame with renamed column:")
print(emp)

# (b)(i) Multiple plot lines
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(emp['Years'], emp['Name'], marker='o')
plt.xlabel('Years')
plt.ylabel('Name')
plt.title('Years vs Name')

plt.subplot(1, 2, 2)
plt.plot(emp['Years'], emp['Gender'], marker='s')
plt.xlabel('Years')
plt.ylabel('Gender')
plt.title('Years vs Gender')
plt.tight_layout()
plt.show()

# (b)(ii) Create subplots
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xlabel('Years')
ax.set_ylabel('Name')
ax.set_title('Age Vs. Name')
plt.show()

# (b)(iii) Line plot with circular markers
plt.figure(figsize=(10, 6))
# Convert names to numerical values for plotting
name_codes = range(len(emp))
plt.plot(emp['Years'], name_codes, marker='o', linestyle='--', color='blue', markersize=8)
plt.yticks(name_codes, emp['Name'])
plt.xlabel('Years')
plt.ylabel('Name')
plt.title('Years vs Names')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
"""
Question 23
"""
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# (a) Import libraries and load dataset
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
print(tips.head())

# (b) Create scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Tip Amount vs Total Bill')
plt.show()

# (c) Display rows where total_bill > 30
high_bills = tips[tips['total_bill'] > 30]
print("Rows with total bill > 30:")
print(high_bills)

# (d) Calculate tip percentage
tips['tip_percentage'] = (tips['tip'] / tips['total_bill']) * 100
print("\nData with tip percentage:")
print(tips[['total_bill', 'tip', 'tip_percentage']].head())

# (e) Display tip percentage statistics
print("\nTip percentage statistics:")
print(tips['tip_percentage'].describe())
```
